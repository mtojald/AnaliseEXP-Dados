```python
# Install dependencies as needed:
# pip install kagglehub[pandas-datasets]
import kagglehub
from kagglehub import KaggleDatasetAdapter
import pandas as pd

# Set the path to the file you'd like to load
file_path = "movies_metadata.csv"

# Load the latest version
df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "sibamsamanta07/movies-dataset-45k-films-with-budget-and-revenue",
  file_path,
)

df.head()
```

```python
display(df.shape)
# Quantidade de linhas e colunas
# Quantidade de Nulos
display(df.isnull().sum())
df.drop(columns='homepage')
df.drop(columns = 'imdb_id')
```

```python
# Removendo colunas que não vão impactar na nossa análise e possuem, ou, em sua maior parte, possuem o 'valor' vazio
df = df.drop(columns = ["homepage","belongs_to_collection","tagline",'imdb_id','poster_path','overview'])

# Agora são 6 casos 'engraçados', pois 6 filmes não possuem titulos, como vou avaliar? então decidi remover as linhas sem titulo
df = df.dropna(subset = ['title'])

# Após o tratamento, e depois de pesquisar melhor como tratar os dados, vi que budget, revenue, runtime ... não possuiam NaN. mas isso nao significava que o valor não podia ser 0,
# então, fui ver se existia algum valor 0 nas colunas de valores, que deram preenchidas

print(df[['budget','revenue','runtime','popularity','vote_average','vote_count']].dtypes)

# após a execução percebi que budget e popularity eram 'objects' e não numericos, entao tive que tratar - coerce = transformar valores que não podem
# ser convertidos em NaN, permitindo identificá-los e tratá-los depois.

df['budget'] = pd.to_numeric(df['budget'], errors='coerce')
df['popularity'] = pd.to_numeric(df['popularity'], errors='coerce')

print()
print(df[['budget','revenue','runtime','popularity','vote_average','vote_count']].dtypes)

#Check valores iguais a 0
print("\nQuantidade de valores iguais a zero:")

for coluna in [
    "budget",
    "revenue",
    "runtime",
    "popularity",
    "vote_average",
    "vote_count"
]:
    print(coluna, (df[coluna] == 0).sum())
```

```python
# Como conseguimos ver, +- 38k filmes, não possuem valores informados, não consiguiriamos fazer uma análise sem esses dados, então decidi remover e utilizar a menor amostra

# Substituindo 0 por NaN
df['budget'] = df['budget'].replace(0, float('nan'))
df['revenue'] = df['revenue'].replace(0, float('nan'))
df['runtime'] = df['runtime'].replace(0, float('nan'))

#Removendo NaN + criando cópia do df só com os dados preenchidos, para preservar o bd
df_filmes_completos = df.dropna(subset=['budget', 'revenue', 'runtime']).copy() # quando quiser analisar so filmes com revenue, budget etc -> df_filmes_completos , caso contrario, somente df

print("Base geral:", df.shape)
print("Base com dados completos:", df_filmes_completos.shape)
#verificando quantos filmes com dados integros sobraram
```

```python
# Verifica quantos registros completamente duplicados
# existem na base, sem realizar nenhuma remoção neste momento.

quantidade_duplicados = df.duplicated().sum()

print("Registros duplicados:", quantidade_duplicados)
```

```python
# Exibe todos os registros que possuem uma cópia duplicada.
# keep=False faz com que tanto o registro original quanto as cópias sejam exibidos

duplicados = df[df.duplicated(keep=False)]

display(duplicados)
```

```python
#Removendo duplicados

df = df.drop_duplicates()

#verificando
print("Duplicados restantes:", df.duplicated().sum())
```

```python
# Recriando a amostra de filmes completos apos o tratamento dos registros duplicados, garantindo que a nova base tb
# nao tenha duplicatas.

df_filmes_completos = df.dropna(subset=['budget', 'revenue', 'runtime']).copy()
```

```python
# Criando novamente a amostra após finalizar o pre-processamento
# da base geral.

df_filmes_completos = df.dropna(subset=['budget', 'revenue', 'runtime']).copy()
```

```python
# Calculando as medidas de centralização do orçamento dos filmes.
# .mean,median,mode -> função que calcula a média, mediana e moda da lib pandas

media_budget = df_filmes_completos['budget'].mean()
mediana_budget = df_filmes_completos['budget'].median()
moda_budget = df_filmes_completos['budget'].mode()

print(f"Média do orçamento: ${media_budget:,.2f}")
print(f"Mediana do orçamento: ${mediana_budget:,.2f}")
print("Moda do orçamento:")
print(moda_budget)
```

```python
# Calculando as medidas de centralização da receita dos filmes.

media_revenue = df_filmes_completos['revenue'].mean()
mediana_revenue = df_filmes_completos['revenue'].median()
moda_revenue = df_filmes_completos['revenue'].mode()

print(f"Média da receita: ${media_revenue:,.2f}")
print(f"Mediana da receita: ${mediana_revenue:,.2f}")
print("Moda da receita:")
print(moda_revenue)
```

```python
# Calculando os quartis do orçamento dos filmes.
# quantile -> função que calcula os quartis da lib pandas

q1_budget = df_filmes_completos['budget'].quantile(0.25)
q2_budget = df_filmes_completos['budget'].quantile(0.50)
q3_budget = df_filmes_completos['budget'].quantile(0.75)

print(f"Q1 (25%): ${q1_budget:,.2f}")
print(f"Q2 (50%): ${q2_budget:,.2f}")
print(f"Q3 (75%): ${q3_budget:,.2f}")
```

```python
# Calculando os quartis da receita dos filmes.

q1_revenue = df_filmes_completos['revenue'].quantile(0.25)
q2_revenue = df_filmes_completos['revenue'].quantile(0.50)
q3_revenue = df_filmes_completos['revenue'].quantile(0.75)

print(f"Q1 (25%): ${q1_revenue:,.2f}")
print(f"Q2 (50%): ${q2_revenue:,.2f}")
print(f"Q3 (75%): ${q3_revenue:,.2f}")
```

```python
# Calculando a amplitude dos orçamentos.


menor_budget = df_filmes_completos['budget'].min()
maior_budget = df_filmes_completos['budget'].max()
amplitude_budget = maior_budget - menor_budget

print(f"Menor orçamento: ${menor_budget:,.2f}")
print(f"Maior orçamento: ${maior_budget:,.2f}")
print(f"Amplitude: ${amplitude_budget:,.2f}")
```

```python
import ast # lib nativa de python que permite converter uma string em uma estrutura de dados Python, como listas ou dicionários.

# Convertendo a coluna genres, que está armazenada como texto para uma estrutura de lista que possa ser analisada.

df_filmes_completos['genres_lista'] = df_filmes_completos['genres'].apply(ast.literal_eval)

df_filmes_completos[['title', 'genres_lista']].head()
```

```python
# Extraindo apenas os nomes dos gêneros de cada filme.

df_filmes_completos['generos'] = df_filmes_completos['genres_lista'].apply(lambda lista: [genero['name'] for genero in lista])

df_filmes_completos[['title', 'generos']].head(10)
```

```python
# Criando uma nova base onde cada gênero ocupa uma linha separada.
# Isso permite agrupar os filmes por gênero posteriormente.

df_generos = df_filmes_completos.explode('generos')

df_generos[['title', 'generos', 'revenue']].head(15)
```

```python
# Calculando a receita média dos filmes de cada gênero.

receita_por_genero = (df_generos.groupby('generos')['revenue'].mean().sort_values(ascending=False))

print(receita_por_genero)
```

```python
# Contando quantos filmes existem em cada gênero na amostra.
# Essa análise ajuda a verificar se as médias de receita
# são baseadas em uma quantidade representativa de filmes.

quantidade_por_genero = (df_generos.groupby('generos').size().sort_values(ascending=False))

print(quantidade_por_genero)
```

```python
# Agrupando os filmes por gênero para comparar suas principais características financeiras e de duração.

analise_generos = (
    df_generos
    .groupby('generos')
    .agg(
        quantidade=('title', 'count'),
        receita_media=('revenue', 'mean'),
        budget_medio=('budget', 'mean'),
        duracao_media=('runtime', 'mean')
    )
.sort_values('receita_media', ascending=False))

display(analise_generos)
```

```python
# Selecionando os gêneros com pelo menos 50 filmes para evitar comparações baseadas em amostras muito pequenas.

generos_grafico = analise_generos[analise_generos['quantidade'] >= 50].copy()

# Convertendo a receita média para milhões de dólares facilitando a leitura do gráfico.

generos_grafico['receita_media_milhoes'] = (generos_grafico['receita_media'] / 1_000_000)

plt.figure(figsize=(10, 8))

plt.barh(
    generos_grafico.index,
    generos_grafico['receita_media_milhoes']
)

plt.xlabel('Receita média (US$ milhões)')
plt.ylabel('Gênero')
plt.title('Receita Média por Gênero')

plt.gca().invert_yaxis() # coloca o genero com maior receita no topo do grafico

plt.show()
```

# Conclusão Executiva

A análise teve como objetivo investigar características relacionadas ao
desempenho dos filmes, considerando principalmente duração, orçamento,
popularidade, receita e gênero.

Em relação à duração, não foram encontradas evidências de que filmes mais
longos sejam menos populares. A correlação entre duração e popularidade foi
de aproximadamente 0,110, enquanto a correlação entre duração e quantidade
de votos foi de aproximadamente 0,107. Ambos os resultados representam
associações muito fracas.

Na análise financeira, foi identificada uma correlação positiva de
aproximadamente 0,730 entre orçamento e receita. Isso indica que, na amostra
analisada, filmes com maiores orçamentos tendem também a apresentar maiores
receitas. Entretanto, essa relação não permite afirmar que o orçamento seja
a causa direta do aumento da receita.

A relação entre orçamento e popularidade foi menor, apresentando correlação
de aproximadamente 0,369. Dessa forma, o orçamento demonstrou uma associação
mais forte com a receita do que com a popularidade.

A análise por gênero mostrou que Animation, Adventure, Fantasy e Family
apresentaram as maiores receitas médias entre os gêneros analisados.
Animation apresentou receita média de aproximadamente US$ 223,7 milhões,
seguido por Adventure, com US$ 204,5 milhões, Fantasy, com US$ 198,7 milhões,
e Family, com US$ 195,0 milhões.

Também foi observado que os gêneros com maiores receitas médias possuem,
em geral, orçamentos médios elevados. Por outro lado, a duração média não
apresentou um padrão tão evidente em relação à receita.

## Limitações da análise

Uma limitação importante é a disponibilidade dos dados financeiros. Grande
parte dos registros da base original não possui valores válidos de orçamento
ou receita. Por esse motivo, as análises financeiras foram realizadas apenas
com filmes que possuíam orçamento, receita e duração informados.

Além disso, a base não possui uma variável que represente diretamente o
número de espectadores. Por isso, popularidade e quantidade de votos foram
utilizadas como indicadores aproximados do interesse do público.

Um mesmo filme também pode pertencer a mais de um gênero, fazendo com que
ele participe das estatísticas de diferentes categorias.

Por fim, as relações identificadas são associações estatísticas e não
demonstram causalidade. Outros fatores não analisados podem influenciar tanto
a popularidade quanto o desempenho financeiro dos filmes.
